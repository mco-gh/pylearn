## Emoji Slot Machine

0–5 min: Setup & pitch

### Symbols & a single spin
- Create SYMBOLS, import random.
- Write spin_reels() with random.choices(..., k=3).
- print(spin_reels()) to see three emojis.

```python
# STEP 1: Random emoji spin

import random

SYMBOLS = ["🍒", "🍋", "🍉", "⭐", "7️⃣"]

def spin_reels():
    # Pick 3 symbols with equal chance
    return random.choices(SYMBOLS, k=3)

def main():
    print("🎰 Emoji Slot Machine – test spin")
    spin = spin_reels()
    print("Result:", " | ".join(spin))

if __name__ == "__main__":
    main()
```

### Scoring
- Add PAYOUTS (dict of triples → coins) and PAIR_PAYOUT.
- Write score(spin) using tuple membership and len(set(spin)).


```python
# STEP 2: Weighted spins + scoring

import random

SYMBOLS = ["🍒", "🍋", "🍉", "⭐", "7️⃣"]
WEIGHTS = [5, 5, 4, 3, 1]  # 7️⃣ is rare!

# Triple matches pay more
PAYOUTS = {
    ("🍒","🍒","🍒"): 5,
    ("🍋","🍋","🍋"): 8,
    ("🍉","🍉","🍉"): 12,
    ("⭐","⭐","⭐"): 20,
    ("7️⃣","7️⃣","7️⃣"): 50
}
PAIR_PAYOUT = 2  # any pair (but not triple)

def spin_reels():
    return random.choices(SYMBOLS, weights=WEIGHTS, k=3)

def score(spin):
    t = tuple(spin)
    if t in PAYOUTS:
        return PAYOUTS[t]
    if len(set(spin)) == 2:  # any pair
        return PAIR_PAYOUT
    return 0

def main():
    print("🎰 Emoji Slot Machine – scoring demo")
    for _ in range(5):
        s = spin_reels()
        print("Spin:", " | ".join(s), "→", score(s), "coins")

if __name__ == "__main__":
    main()
```

### Game loop & coins
- Add constants START_COINS, SPIN_COST.
- While loop: ask for input, deduct cost, spin, compute winnings, update coins.
- End when coins run out or user quits.


```python
# STEP 3: Add a game loop with coins

import random

SYMBOLS = ["🍒", "🍋", "🍉", "⭐", "7️⃣"]
WEIGHTS = [5, 5, 4, 3, 1]

PAYOUTS = {
    ("🍒","🍒","🍒"): 5,
    ("🍋","🍋","🍋"): 8,
    ("🍉","🍉","🍉"): 12,
    ("⭐","⭐","⭐"): 20,
    ("7️⃣","7️⃣","7️⃣"): 50
}
PAIR_PAYOUT = 2

SPIN_COST   = 1
START_COINS = 10

def spin_reels():
    return random.choices(SYMBOLS, weights=WEIGHTS, k=3)

def score(spin):
    t = tuple(spin)
    if t in PAYOUTS:
        return PAYOUTS[t]
    if len(set(spin)) == 2:
        return PAIR_PAYOUT
    return 0

def main():
    print("🎰 EMOJI SLOTS – any pair wins 2, triples pay more!")
    print(f"Spins cost {SPIN_COST}. You start with {START_COINS} coins.\n")

    coins = START_COINS
    while coins >= SPIN_COST:
        choice = input("[Enter] spin  |  (q) quit\n> ").strip().lower()
        if choice == "q":
            break

        coins -= SPIN_COST
        final = spin_reels()
        print("🎲", " | ".join(final))

        won = score(final)
        if won:
            coins += won
            print(f"🎉 You won {won} coins!")
        else:
            print("🙃 No win this time.")
        print(f"💰 Coins: {coins}\n")

    print("\nOut of coins or you quit. Thanks for playing!")

if __name__ == "__main__":
    main()
```

### Animation” polish
- Add draw_frame() and a tiny loop with time.sleep(..) and cursor moves to simulate spinning.


```python
# STEP 4: Add a tiny animation effect while spinning

import random, time, sys

SYMBOLS = ["🍒", "🍋", "🍉", "⭐", "7️⃣"]
WEIGHTS = [5, 5, 4, 3, 1]

PAYOUTS = {
    ("🍒","🍒","🍒"): 5,
    ("🍋","🍋","🍋"): 8,
    ("🍉","🍉","🍉"): 12,
    ("⭐","⭐","⭐"): 20,
    ("7️⃣","7️⃣","7️⃣"): 50
}
PAIR_PAYOUT = 2

SPIN_COST   = 1
START_COINS = 10

def spin_reels(rigged=None):
    if rigged:
        return list(rigged)
    return random.choices(SYMBOLS, weights=WEIGHTS, k=3)

def score(spin):
    t = tuple(spin)
    if t in PAYOUTS:
        return PAYOUTS[t]
    if len(set(spin)) == 2:
        return PAIR_PAYOUT
    return 0

def draw_frame(spin, coins, flicker=False):
    # If flicker, randomize middle symbol for a “spin” feel
    shown = spin.copy()
    if flicker:
        shown[1] = random.choice(SYMBOLS)
    reels = " | ".join(shown)
    print("+-------------------+")
    print(f"|  {reels}  |   💰 {coins}")
    print("+-------------------+")

def main():
    print("🎰 EMOJI SLOTS – animation demo")
    print(f"Spins cost {SPIN_COST}. Start: {START_COINS} coins.\n")

    coins = START_COINS
    while coins >= SPIN_COST:
        choice = input("[Enter] spin  |  (q) quit\n> ").strip().lower()
        if choice == "q":
            break

        coins -= SPIN_COST
        final = spin_reels()

        # Quick “spinning” animation: redraw same 3-line box, flickering middle
        for _ in range(10):
            draw_frame(final, coins, flicker=True)
            time.sleep(0.06)
            sys.stdout.write("\x1b[3A")  # move cursor up 3 lines
        sys.stdout.write("\x1b[3B")      # move back down
        draw_frame(final, coins, flicker=False)

        won = score(final)
        if won:
            coins += won
            print(f"🎉 You won {won} coins!")
        else:
            print("🙃 No win this time.")
        print(f"Coins: {coins}\n")

    print("\nOut of coins or you quit. Thanks for playing!")

if __name__ == "__main__":
    main()
```

### High score + Easter egg
- File I/O: load_high_score()/save_high_score().
- Track high, congratulate on new record.
- Add a playful teacher demo: typing jackpot rigs the next spin to 7️⃣7️⃣7️⃣.


```python
# STEP 5: High score persistence + secret 'jackpot' rig

import random, time, sys

SYMBOLS = ["🍒", "🍋", "🍉", "⭐", "7️⃣"]
WEIGHTS = [5, 5, 4, 3, 1]

PAYOUTS = {
    ("🍒","🍒","🍒"): 5,
    ("🍋","🍋","🍋"): 8,
    ("🍉","🍉","🍉"): 12,
    ("⭐","⭐","⭐"): 20,
    ("7️⃣","7️⃣","7️⃣"): 50
}
PAIR_PAYOUT = 2

SPIN_COST   = 1
START_COINS = 10

def spin_reels(rigged=None):
    if rigged:
        return list(rigged)
    return random.choices(SYMBOLS, weights=WEIGHTS, k=3)

def score(spin):
    t = tuple(spin)
    if t in PAYOUTS:
        return PAYOUTS[t]
    if len(set(spin)) == 2:
        return PAIR_PAYOUT
    return 0

def draw_frame(spin, coins, flicker=False):
    shown = spin.copy()
    if flicker:
        shown[1] = random.choice(SYMBOLS)
    reels = " | ".join(shown)
    print("+-------------------+")
    print(f"|  {reels}  |   💰 {coins}")
    print("+-------------------+")

def load_high_score():
    try:
        with open("highscore.txt", "r", encoding="utf-8") as f:
            return int(f.read().strip() or "0")
    except Exception:
        return 0

def save_high_score(score_value):
    try:
        with open("highscore.txt", "w", encoding="utf-8") as f:
            f.write(str(score_value))
    except Exception:
        pass

def main():
    print("🎰  EMOJI SLOTS  🎰")
    print("Match three for big prizes. Any pair wins 2 coins.")
    print("7️⃣7️⃣7️⃣ pays 50! Spins cost 1 coin.\n")

    coins = START_COINS
    high = load_high_score()
    rig_next = None

    while coins >= SPIN_COST:
        choice = input("[Enter] spin  |  (q) quit  |  (h) high score  |  (✨) secret: type 'jackpot'\n> ").strip().lower()
        if choice == "q":
            break
        if choice == "h":
            print(f"🏆 High score: {high} coins\n")
            continue
        if choice == "jackpot":
            rig_next = ("7️⃣","7️⃣","7️⃣")

        coins -= SPIN_COST
        final = spin_reels(rigged=rig_next)
        rig_next = None

        for _ in range(10):
            draw_frame(final, coins, flicker=True)
            time.sleep(0.06)
            sys.stdout.write("\x1b[3A")
        sys.stdout.write("\x1b[3B")
        draw_frame(final, coins, flicker=False)

        won = score(final)
        if won:
            coins += won
            print(f"🎉 You won {won} coins!")
        else:
            print("🙃 No win this time.")
        print(f"Coins: {coins}\n")

        if coins > high:
            high = coins
            save_high_score(high)
            print("🏅 New high score!\n")

    print("\nGame over. Thanks for playing!")
    print(f"Your high score: {high}")

if __name__ == "__main__":
    main()
```

### Step 3 — Real spirograph (hypotrochoid curve)

Hypotrochoid: a point on a small circle rolling inside a big circle.

Formulas (t is angle in radians):
- x = (R−r)·cos t + d·cos(((R−r)/r)·t)
- y = (R−r)·sin t − d·sin(((R−r)/r)·t)

```python
# STEP 3: Hypotrochoid spirograph with basic parameters

import turtle as T
import math

# Parameters (tweak these!):
R = 150    # big (fixed) circle radius
r = 60     # small (rolling) circle radius
d = 70     # pen offset from small-circle center

screen = T.Screen()
screen.title("Hypotrochoid")
screen.colormode(255)
pen = T.Turtle()
pen.speed(0)
pen.pensize(2)
pen.color(20, 120, 240)

# Faster drawing: manual screen updates
screen.tracer(0)

def point(t):
    k = (R - r)
    x = k * math.cos(t) + d * math.cos(k / r * t)
    y = k * math.sin(t) - d * math.sin(k / r * t)
    return x, y

# Compute how long to draw: LCM-like using multiples of pi
# Draw enough turns for the curve to close:
turns = r // math.gcd(int(R), int(r))
steps = 2000

# Move to start
x0, y0 = point(0.0)
pen.up(); pen.goto(x0, y0); pen.down()

for i in range(steps * turns):
    t = 2 * math.pi * i / steps
    x, y = point(t)
    pen.goto(x, y)
    if i % 200 == 0:
        screen.update()

screen.update()
screen.mainloop()
```

### Step 4 — Color cycling (rainbow!) + helpers

```python
# STEP 4: Rainbow color cycling, smoother animation

import turtle as T
import math

screen = T.Screen()
screen.title("Rainbow Hypotrochoid")
screen.colormode(255)
pen = T.Turtle()
pen.speed(0); pen.pensize(2)

# Parameters to play with
R, r, d = 160, 51, 100

# Manual updates for speed
screen.tracer(0)

def hsv_to_rgb(h, s=1.0, v=1.0):
    """h in [0,1) -> (r,g,b) in 0..255"""
    i = int(h*6)
    f = h*6 - i
    p = int(255*v*(1-s))
    q = int(255*v*(1-f*s))
    t = int(255*v*(1-(1-f)*s))
    v = int(255*v)
    i %= 6
    return [
        (v,t,p),(q,v,p),(p,v,t),(p,q,v),(t,p,v),(v,p,q)
    ][i]

def pt(t):
    k = (R - r)
    x = k*math.cos(t) + d*math.cos(k/r * t)
    y = k*math.sin(t) - d*math.sin(k/r * t)
    return x, y

# How long to draw (close the curve)
turns = r // math.gcd(int(R), int(r))
steps = 2400

x0, y0 = pt(0.0)
pen.up(); pen.goto(x0, y0); pen.down()

for i in range(steps * turns):
    t = 2*math.pi * i / steps
    x, y = pt(t)
    # set a smooth rainbow color
    h = (i % steps) / steps
    pen.pencolor(hsv_to_rgb(h))
    pen.goto(x, y)
    if i % 300 == 0:  # fewer updates = faster
        screen.update()

screen.update()
screen.mainloop()
```

### Step 5 — Interactive spiro lab (inputs, keys, save)
- Inputs: width/height, R, r, d, and speed.
- Keys:
  - Space = pause/resume
  - N = random new pattern
  - S = save PostScript (EPS-like) to spiro.ps

```python
# STEP 5: Interactive Spirograph Lab (inputs + keys + save)

import turtle as T
import math, random

# ---------- UI helpers ----------
def ask_int(label, default, lo, hi):
    s = screen.textinput("Input", f"{label} [{default}]: ")
    if not s: return default
    try:
        v = int(s)
        return max(lo, min(hi, v))
    except:
        return default

def hsv_to_rgb(h, s=1.0, v=1.0):
    i = int(h*6); f = h*6 - i
    p = int(255*v*(1-s))
    q = int(255*v*(1-f*s))
    t = int(255*v*(1-(1-f)*s))
    v = int(255*v)
    i %= 6
    return [(v,t,p),(q,v,p),(p,v,t),(p,q,v),(t,p,v),(v,p,q)][i]

# ---------- Setup ----------
screen = T.Screen()
screen.title("Interactive Spirograph Lab")
screen.colormode(255)

# Window size
w = 900; h = 700
try:
    screen.setup(w, h)
except:
    pass

pen = T.Turtle(visible=True)
pen.speed(0); pen.pensize(2)

paused = False

def draw_spiro(R, r, d, steps=2600, update_every=400, rainbow=True):
    screen.tracer(0)
    pen.clear()
    pen.up()

    def pt(t):
        k = (R - r)
        return (
            k*math.cos(t) + d*math.cos(k/r * t),
            k*math.sin(t) - d*math.sin(k/r * t),
        )

    turns = max(1, r // math.gcd(int(R), int(r)))
    x0, y0 = pt(0.0)
    pen.goto(x0, y0); pen.down()

    for i in range(steps * turns):
        if paused:
            screen.update()
            screen.ontimer(lambda: None, 50)
            continue
        t = 2*math.pi * i / steps
        x, y = pt(t)
        if rainbow:
            pen.pencolor(hsv_to_rgb((i % steps) / steps))
        pen.goto(x, y)
        if i % update_every == 0:
            screen.update()
    screen.update()

# ---------- Controls ----------
def toggle_pause():
    global paused
    paused = not paused

def save_ps():
    # Saves vector PostScript you can convert to PDF/PNG later
    cv = screen.getcanvas()
    cv.postscript(file="spiro.ps")
    print("Saved to spiro.ps")

def random_pattern():
    # Fun ranges that usually look good
    R  = random.randint(120, 220)
    r  = random.randint(20,  120)
    d  = random.randint(20,  160)
    draw_spiro(R, r, d)

screen.listen()
screen.onkey(toggle_pause, "space")
screen.onkey(save_ps, "s")
screen.onkey(random_pattern, "n")

# ---------- First run with inputs ----------
R = ask_int("Big radius R (80-260)", 180, 80, 260)
r = ask_int("Small radius r (10-140)", 55, 10, 140)
d = ask_int("Pen offset d (10-180)", 90, 10, 180)
draw_spiro(R, r, d)

screen.mainloop()
```
